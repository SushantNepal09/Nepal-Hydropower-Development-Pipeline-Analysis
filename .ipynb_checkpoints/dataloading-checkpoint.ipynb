```python
import pandas as pd
```

```python
survey_url = "https://doed.gov.np/pages/hydromorethan1/"
construction_url = "https://doed.gov.np/pages/clhydromorethan1/"
operating_url = "https://doed.gov.np/pages/powerplantsmorethan1/"
```

```python
operating = pd.read_html(operating_url)[0]
construction = pd.read_html(construction_url)[0]
survey = pd.read_html(survey_url)[0]
```

```python
print(operating.head())
print(construction.head())
print(survey.head())
```

```python
operating.to_csv("operating.csv", index = False)
construction.to_csv("construction.csv",index= False)
survey.to_csv("survey.csv",index = False)
```
