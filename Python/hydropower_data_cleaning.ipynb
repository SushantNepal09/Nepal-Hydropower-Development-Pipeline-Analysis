```python
from bikram_sambat import date
import pandas as pd

def convert_bs_to_ad(date_str):
    if pd.isna(date_str) or str(date_str).strip() == "":
        return None

    try:
        y, m, d = map(int, str(date_str).strip().split("-"))

        bs_date = date(y, m, d)
        ad_date = bs_date.togregorian()

        return ad_date.strftime("%Y-%m-%d")

    except Exception:
        return None


dfsur["issue_date_ad"] = dfsur["issue_date"].apply(convert_bs_to_ad)
```

```python
def dms_to_decimal(dms):
    if pd.isna(dms):
        return None
    dms = dms.replace('o', ' ').replace('"', '').replace("'", ' ')
    degrees, minutes, seconds = map(float, dms.split())
    return degrees + minutes / 60 + seconds / 3600
```
